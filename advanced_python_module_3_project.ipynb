```python
import pandas as pd
```

```python
import requests
from bs4 import BeautifulSoup
```

```python
response = requests.get('https://btech-data-analytics.github.io/bridgerland-technical-college/home.html')
```

```python
soup = BeautifulSoup(response.text, 'html.parser')
```

```python
rows = soup.find_all('div', class_='w3-container')
```

```python
headers = ['Title', 'Author']
```

```python
data = []

# Find all book cards within the 'Featured Books' section (rows[1])
books = rows[1].find_all('div', class_='w3-card-4')

for card in books:
    # Extract title and author from each card
    title_tag = card.find('h3')
    author_tag = card.find('p')

    title = title_tag.text.strip() if title_tag else ''
    author = author_tag.text.strip() if author_tag else ''

    data.append([title, author])
```

```python
books = pd.DataFrame(data, columns=headers)
```
