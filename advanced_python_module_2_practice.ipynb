```python
import matplotlib.pyplot as plt
import seaborn as sns
import pandas as pd
```

```python
df = pd.read_csv("companies.csv")
```

```python
df.head()
```

```python
plt.figure(figsize=(12,6))

sns.countplot(data=df, x='state', order=df['state'].value_counts().index)

plt.title("Number of Companies by State")
plt.xticks(rotation=90)

plt.show()
```

```python
industry_jobs = df.groupby('Industry')['job_openings'].sum().sort_values(ascending=False)

plt.figure(figsize=(12,6))

sns.barplot(
    x=industry_jobs.index,
    y=industry_jobs.values
)

plt.title("Total Job Openings by Industry")
plt.xticks(rotation=90)

plt.show()
```

```python
growth = df.groupby('founded')['growth_percentage'].mean().sort_values(ascending=False)

plt.figure(figsize=(12,6))

sns.barplot(
    x=growth.index,
    y=growth.values
)

plt.title("Average growth per year")
plt.xlabel("Year")
plt.ylabel("Average Growth")
plt.xticks(rotation=90)

plt.show()
```

```python
plt.figure(figsize=(12,6))

sns.scatterplot(
    data=df,
    x='employees',
    y='estimated_revenues'
)

plt.title("Estimated Revenues by Number of Employees")
plt.xlabel("Employees")
plt.ylabel("Estimated Revenues")

plt.show()
```

```python
plt.figure(figsize=(12,6))

sns.histplot(
    data=df,
    x='founded',
    bins=20
)

plt.title("Number of companies founded each year")
plt.xlabel("Founded Year")
plt.ylabel("Number of Companies")

plt.show()
```
