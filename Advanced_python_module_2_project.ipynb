```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
df = pd.read_csv('housing.csv')
```

```python
df.head()
```

```python
plt.figure(figsize=(12,6))

sns.scatterplot(
    data=df,
    x='Landsize',
    y='Price',
    hue='Type'
)

plt.xlabel("Landsize")
plt.ylabel("price")

plt.show()
```

```python
avg_price = df.groupby('Date')['Price'].mean()
```

```python
df.head()
```

```python
plt.figure(figsize=(12,6))

avg_price_by_year = df.groupby('YearBuilt')['Price'].mean()

sns.lineplot(
    x=avg_price_by_year.index,
    y=avg_price_by_year.values
)
plt.title('average house price by year built')
plt.xlabel('year built')
plt.ylabel('Price')
plt.xticks(rotation=90)
plt.show()
```

```python
growth = df.groupby('Regionname')['Price'].mean().sort_values(ascending=False)

plt.figure(figsize=(12,6))

sns.barplot(
    x=growth.index,
    y=growth.values
)

plt.title("price per region")
plt.xlabel("region")
plt.ylabel("Average price")
plt.xticks(rotation=90)

plt.show()
```

```python
plt.figure(figsize=(12,6))

sns.histplot(
    data=df,
    x='Price',
    bins=20
)

plt.title("Property Prices")
plt.xlabel("Price")
plt.ylabel("Frequency")

plt.show()
```

```python
plt.figure(figsize=(12,6))

sns.boxplot(
    data=df,
    x='YearBuilt',
    y='Regionname',
)

plt.title("Year built by region")
plt.xlabel("Year Built")
plt.ylabel("Region")
plt.xticks(rotation=90)

plt.show()
```
